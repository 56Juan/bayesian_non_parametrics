# 11_05 — Modelos de referencia · Escenario 1

Paso **3c** del ciclo, posterior a `11_04_evaluacion`. Responde una sola
pregunta del eje 1 de `docs/03 Modelo.tex §03_06`:

> ¿el PSBPM-FD supera a los modelos de referencia **sobre la misma
> representación**?

## Alcance, y lo que queda deliberadamente fuera

| Se reporta | No se reporta |
|---|---|
| Métricas **puntuales**: RMSE y $R^2$ por score, MISE y RMSE funcional | CRPS, energy score, cobertura, PIT |
| Evolución del MISE sobre ventana móvil | Calibración condicional |
| Piso de truncamiento FPCA como referencia común | Probabilidades de inclusión |

La restricción es de fondo, no de tiempo. Ninguno de los competidores entrega
una predictiva completa comparable con la del PSBPM-FD: el VAR y el FAR sólo
la tienen bajo un supuesto gaussiano que el estudio existe para cuestionar, y
los modelos de aprendizaje automático no la tienen en absoluto. Enfrentarlos
en CRPS exigiría inventarles una ley predictiva, y la comparación mediría esa
invención. La **media condicional**, en cambio, está bien definida para los
cinco, y es lo único que se compara aquí.

## Las tres reglas que hacen legítima la comparación

1. **Misma representación.** Todos operan sobre los mismos scores FPCA, con la
   base y el estandarizador ajustados *sólo* en entrenamiento y cargados desde
   los mismos artefactos que usó el PSBPM-FD. Ningún competidor reajusta el
   FPCA.
2. **Mismo conjunto de información.** Todos reciben la matriz de diseño de
   `dataset_fpc_*`, es decir los `N_LAGS` rezagos de las `M` componentes. La
   comparación es entre **formas funcionales**, no entre información
   disponible.
3. **Sin reentrenamiento.** Cada modelo se estima una vez con los orígenes de
   entrenamiento y predice a $h=1$ sobre la serie completa, igual que el
   PSBPM-FD. El salto en $T_0$ es entonces comparable entre modelos.

## Los competidores

| Modelo | Qué es | Referencia |
|---|---|---|
| `VAR` | VAR($p$) sobre los scores estandarizados, con intercepto, por mínimos cuadrados | Aue, Norinho & Hörmann (2015) |
| `FAR1` | $\hat\Psi=\hat C_1\hat C_0^{-1}$ sobre las curvas, proyectado en el subespacio FPCA | Bosq (2000) |
| `RF` | Bosque aleatorio multisalida sobre la misma matriz de diseño | competidor no lineal |
| `GBT` | *Gradient boosting* por score | competidor no lineal |
| `media`, `persistencia` | las dos líneas base de `fit/baselines.py` | piso |

El PSBPM-FD **no se recalcula**: sus cifras se leen de los CSV que escribió
`11_04`. Recomputar la predictiva exigiría releer las trazas y volver a
propagarlas, y el resultado sería idéntico por construcción.

> **Nota sobre `VAR` y `FAR1`.** Con $p=$ `N_LAGS` $=1$ y truncamiento $K=M$
> ambos estiman el mismo operador lineal por caminos distintos: el VAR por MCO
> sobre scores estandarizados y con intercepto, el FAR por operadores de
> covarianza sobre scores en escala original y sin intercepto. Que sus cifras
> casi coincidan **es el resultado esperado** y sirve de control cruzado; la
> diferencia entre ambas es exactamente lo que aportan el intercepto y el
> centrado. Lo que el FAR sí entrega y el VAR no es $\|\hat\Psi\|_{HS}$,
> directamente comparable con el `0.7` que usó el generador del Escenario 1.

```python
# Curvas de referencia alineadas a los origenes, y el piso de la representacion.
X_true_ev = X_true[N_LAGS:]                       # (n_orig, G) VERDADERA
X_proj    = fpca.reconstruct(SCORES)[N_LAGS:]     # proyeccion FPCA sin modelo


def curvas_desde_scores_std(Y_std_pred):
    """Prediccion en scores estandarizados -> curvas en la grilla (n, G).

    Es el mapa determinista de `PropagadorFuncional` con modo_residuo="ninguno":
    des-estandarizar y reconstruir. Se escribe aqui y no se instancia el
    propagador porque este devuelve extracciones de una predictiva y aqui solo
    hay una media condicional.
    """
    S = scores_standardizer.inverse_transform(np.atleast_2d(Y_std_pred))
    return fpca.reconstruct(S)


# Control: reconstruir el objetivo observado debe dar exactamente X_proj.
assert np.allclose(curvas_desde_scores_std(Y_obs), X_proj, atol=1e-8), (
    "El mapa scores->curva no reproduce la proyeccion FPCA; revisar el "
    "estandarizador cargado.")

MISE_TRUNC = {b: mise(X_true_ev[m], X_proj[m], grilla)
              for b, m in (("train", es_train), ("test", ~es_train))}
print("MISE del TRUNCAMIENTO FPCA (piso comun a todos los modelos):")
for b, v in MISE_TRUNC.items():
    print(f"  {b:5s}: {v:.6f}")
print("  <- ningun modelo sobre esta representacion puede bajar de ahi")
```

## 3. Los competidores

Cada uno se ajusta **sólo con los orígenes de entrenamiento** y devuelve una
matriz `(n_orig, M)` de predicciones a $h=1$ en escala estandarizada. Esa firma
común es lo que permite evaluarlos a todos con el mismo código en §4.

### 3.1 VAR($p$) sobre los scores — Aue, Norinho & Hörmann (2015)

FPCA seguido de un VAR sobre el vector de scores. Con $p=N_{lags}$ el regresor
es la propia matriz de diseño del PSBPM-FD, de modo que el VAR es **el caso
lineal-gaussiano anidado** en el modelo propuesto: un PSBP con un único átomo y
pesos constantes se reduce a esto. La diferencia de desempeño entre ambos es,
por tanto, exactamente lo que aporta la mezcla.

El orden se deja en `N_LAGS` para no alterar el conjunto de información. Lo que
sí se reporta es el orden que el BIC habría elegido con libertad, como
diagnóstico: si el BIC pide más rezagos, la limitación es del diseño del
estudio y no del VAR, y hay que declararlo.

```python
def ajustar_var(Z, Y, mask_train):
    """VAR(p) por MCO con intercepto, estimado solo en `mask_train`.

    Se resuelve con `lstsq` y no con la ecuacion normal: con M=4 y p pequeno el
    condicionamiento no es problema, pero la forma normal eleva al cuadrado el
    numero de condicion sin ninguna ganancia.
    """
    Zc = np.hstack([np.ones((Z.shape[0], 1)), Z])
    coef, *_ = np.linalg.lstsq(Zc[mask_train], Y[mask_train], rcond=None)
    return Zc @ coef, coef


PRED = {}          # {nombre: (n_orig, M) en escala estandarizada}
INFO = {}          # notas por modelo, para el resumen

PRED["VAR"], coef_var = ajustar_var(Z, Y_obs, es_train)
INFO["VAR"] = f"p={N_LAGS}, {coef_var.size} parametros"
print(f"VAR({N_LAGS}) ajustado con {int(es_train.sum())} origenes · "
      f"{coef_var.shape[0]} coef. x {coef_var.shape[1]} ecuaciones")
```

```python
# [DIAGNOSTICO] orden que el BIC elegiria con libertad, sobre la serie de
# scores completa. No se usa para predecir: solo dice si N_LAGS deja fuera
# dependencia que el VAR podria aprovechar.
def bic_var(SCORES_STD, T0, p):
    S = np.asarray(SCORES_STD, dtype=float)
    T_, M_ = S.shape
    Yp = S[p:]
    Zp = np.hstack([S[p - j - 1:T_ - j - 1] for j in range(p)])
    Zp = np.hstack([np.ones((Zp.shape[0], 1)), Zp])
    tr = np.arange(p, T_) < T0
    coef, *_ = np.linalg.lstsq(Zp[tr], Yp[tr], rcond=None)
    E = Yp[tr] - Zp[tr] @ coef
    n_ = E.shape[0]
    Sig = E.T @ E / n_
    signo, logdet = np.linalg.slogdet(Sig)
    return n_ * logdet + np.log(n_) * coef.size


P_MAX = 5
_bics = {p: bic_var(SCORES_STD, T0, p) for p in range(1, P_MAX + 1)}
_p_bic = min(_bics, key=_bics.get)
print(f"{'p':>3}  {'BIC':>12}")
for p, v in _bics.items():
    print(f"{p:>3}  {v:>12.2f}" + ("   <- minimo" if p == _p_bic else ""))
print(f"\nBIC elige p={_p_bic}; el estudio fija N_LAGS={N_LAGS}."
      + ("  OK coinciden" if _p_bic == N_LAGS else
         "  <- declarar esta restriccion al reportar"))
```

### 3.2 FAR(1) sobre la curva — Bosq (2000)

$\hat\Psi=\hat C_1\hat C_0^{-1}$, con los operadores de covarianza estimados en
entrenamiento y regularizados por proyección sobre las $M$ autofunciones
retenidas. En esas coordenadas —ortonormales en $L^2$ por construcción del
FPCA— el operador es una matriz $M\times M$ y su norma de Hilbert-Schmidt es la
norma de Frobenius de esa matriz, directamente comparable con la del generador.

A diferencia del VAR opera sobre los scores **en escala original y sin
intercepto**, que es la especificación del modelo autorregresivo hilbertiano.

```python
RIDGE_FAR = 0.0   # [CONFIG] regularizacion de C0. Con M pequeno no hace falta.


def ajustar_far1(SCORES, T0, N_LAGS, ridge=0.0):
    """FAR(1) por Yule-Walker en el subespacio FPCA.

    Devuelve (Xi_pred, Psi_hat) con Xi_pred alineado a los origenes
    t = N_LAGS+1 .. T, en escala ORIGINAL de scores.
    """
    S = np.asarray(SCORES, dtype=float)
    T_ = S.shape[0]
    Xi_lag = S[N_LAGS - 1:T_ - 1]        # valor en t-1 de cada origen
    Xi_now = S[N_LAGS:]                  # valor en t
    tr = np.arange(N_LAGS + 1, T_ + 1) <= T0

    A, Bm = Xi_lag[tr], Xi_now[tr]
    n_ = A.shape[0]
    C0 = A.T @ A / n_
    C1 = Bm.T @ A / n_
    if ridge > 0:
        C0 = C0 + ridge * np.trace(C0) / C0.shape[0] * np.eye(C0.shape[0])
    Psi_hat = np.linalg.solve(C0.T, C1.T).T      # C1 @ inv(C0), sin invertir
    return Xi_lag @ Psi_hat.T, Psi_hat


Xi_far, Psi_hat = ajustar_far1(SCORES, T0, N_LAGS, ridge=RIDGE_FAR)
PRED["FAR1"] = scores_standardizer.transform(Xi_far)

_hs   = float(np.linalg.norm(Psi_hat, "fro"))
_rho  = float(np.max(np.abs(np.linalg.eigvals(Psi_hat))))
_cond = float(np.linalg.cond(np.cov(SCORES[N_LAGS - 1:T0 - 1].T)))
INFO["FAR1"] = f"K={M_fpca}, ||Psi||_HS={_hs:.4f}"

print(f"||Psi_hat||_HS   = {_hs:.4f}    (generador del Escenario 1: 0.70)")
print(f"radio espectral  = {_rho:.4f}   "
      + ("OK contractivo" if _hs < 1 else "<- NO contractivo"))
print(f"cond(C0)         = {_cond:.1f}")
print("\nEl estimador trunca en K=M, de modo que ||Psi_hat||_HS mide el operador")
print("RESTRINGIDO al subespacio retenido: subestima la norma del generador en")
print("la parte de energia que las M componentes no capturan.")
```

### 3.3 Modelos de aprendizaje automático

Dos, y ninguno afinado por validación cruzada sobre el bloque de prueba: eso
sería exactamente la fuga que el resto del diseño evita. Los hiperparámetros
son valores por defecto razonables, declarados aquí y fijos entre escenarios
por la misma razón por la que lo están los del PSBPM-FD.

Su papel en el estudio es acotar cuánto de la mejora del modelo propuesto es
**no linealidad** y cuánto es la **mezcla**. En el Escenario 1 —FAR(1) lineal
gaussiano— lo esperable es que **pierdan** contra el VAR: no hay no linealidad
que aprovechar y el costo es varianza de estimación. Que ganen aquí sería señal
de un problema, no de un acierto.

```python
SEED_ML = 20260830

# [CONFIG] Hiperparametros fijos, sin ajuste por validacion cruzada.
#   min_samples_leaf alto y max_leaf_nodes moderado porque n_train ~ 280
#   origenes: con arboles profundos el sobreajuste se ve en train y no dice
#   nada sobre el test.
PARAMS_RF  = dict(n_estimators=500, min_samples_leaf=5, max_features="sqrt",
                  random_state=SEED_ML, n_jobs=-1)
PARAMS_GBT = dict(max_iter=300, learning_rate=0.05, max_leaf_nodes=15,
                  min_samples_leaf=10, early_stopping=False,
                  random_state=SEED_ML)

rf = RandomForestRegressor(**PARAMS_RF).fit(Z[es_train], Y_obs[es_train])
PRED["RF"] = rf.predict(Z)
INFO["RF"] = (f"n_estimators={PARAMS_RF['n_estimators']}, "
              f"leaf>={PARAMS_RF['min_samples_leaf']}")

# HistGradientBoosting es univariante: un modelo por score, misma Z.
pred_gbt = np.empty_like(Y_obs)
for k in range(n_components):
    g = HistGradientBoostingRegressor(**PARAMS_GBT)
    g.fit(Z[es_train], Y_obs[es_train, k])
    pred_gbt[:, k] = g.predict(Z)
PRED["GBT"] = pred_gbt
INFO["GBT"] = (f"max_iter={PARAMS_GBT['max_iter']}, "
               f"lr={PARAMS_GBT['learning_rate']}")

print(f"RF  ajustado: {int(es_train.sum())} origenes x {p_dis} covariables (multisalida)")
print(f"GBT ajustado: {n_components} modelos univariantes sobre la misma Z")
```

### 3.4 Las dos líneas base

`fit/baselines.py` las evalúa sólo sobre el bloque de prueba; aquí se
reconstruyen sobre la serie completa de orígenes para que entren en la misma
tabla y en la misma ventana móvil. La media incondicional es idénticamente
cero por estar los scores estandarizados con los momentos de entrenamiento, y
por eso su RMSE $\approx 1$ es la unidad de lectura de toda la tabla.

```python
PRED["media"] = np.zeros_like(Y_obs)
INFO["media"] = "cero por construccion en escala estandarizada"

# Persistencia: el score observado en t-1. Para el primer origen (t = N_LAGS+1)
# el valor en t-1 = N_LAGS existe en SCORES_STD, de modo que no hay hueco.
PRED["persistencia"] = SCORES_STD[N_LAGS - 1:T - 1]
INFO["persistencia"] = "y_hat(t) = y(t-1)"

assert all(v.shape == Y_obs.shape for v in PRED.values()), {
    k: v.shape for k, v in PRED.items()}
print("modelos listos:", ", ".join(PRED))
```

## 4. Métricas puntuales, entrenamiento contra prueba

Dos niveles, los mismos de `11_04 §4`:

- **Scores** — RMSE, $R^2$ y razón de dispersión por componente, en escala
  estandarizada. El RMSE de la media incondicional es la unidad.
- **Curva** — MISE y RMSE funcional contra la curva **verdadera**, con
  `modo_residuo="ninguno"`: lo que queda fuera es truncamiento FPCA puro, y ese
  piso aparece en la tabla como fila propia.

```python
def metricas_modelo(nombre, Y_pred):
    """Fila por bloque con las metricas puntuales de los dos niveles."""
    filas = []
    X_pred = curvas_desde_scores_std(Y_pred)
    for etq, mask in (("train", es_train), ("test", ~es_train)):
        y, q = Y_obs[mask], Y_pred[mask]
        filas.append({
            "modelo": nombre, "bloque": etq,
            "RMSE_scores": float(rmse_por_coeficiente(y, q).mean()),
            "R2_scores":   float(np.nanmean(r2_por_columna(y, q, centrar=True))),
            "sd_ratio":    float(np.nanmean(razon_dispersion(y, q))),
            "MISE":        mise(X_true_ev[mask], X_pred[mask], grilla),
            "RMSE_func":   rmse_funcional(X_true_ev[mask], X_pred[mask], grilla),
            "n": int(mask.sum()),
        })
    return filas


comp_df = pd.DataFrame([f for nom, Yp in PRED.items()
                        for f in metricas_modelo(nom, Yp)])

# El piso de truncamiento entra como una fila mas: no es un modelo, es la cota.
comp_df = pd.concat([comp_df, pd.DataFrame([
    {"modelo": "(truncamiento FPCA)", "bloque": b, "RMSE_scores": np.nan,
     "R2_scores": np.nan, "sd_ratio": np.nan, "MISE": MISE_TRUNC[b],
     "RMSE_func": rmse_funcional(X_true_ev[m], X_proj[m], grilla),
     "n": int(m.sum())}
    for b, m in (("train", es_train), ("test", ~es_train))])], ignore_index=True)

comp_df = comp_df.set_index(["bloque", "modelo"]).sort_index()
comp_df.to_csv(PATHS["out_report"] / f"{PREFIJO}_comparacion_modelos.csv")

display(comp_df.style
        .format({"RMSE_scores": "{:.4f}", "R2_scores": "{:.4f}",
                 "sd_ratio": "{:.4f}", "MISE": "{:.6f}", "RMSE_func": "{:.6f}"},
                na_rep="—")
        .background_gradient(subset=["RMSE_scores"], cmap="RdYlGn_r")
        .background_gradient(subset=["MISE"], cmap="RdYlGn_r")
        .set_caption("Metricas PUNTUALES por modelo y bloque · MISE contra la "
                     "curva VERDADERA"))
```

```python
# RMSE por componente: el promedio de la tabla anterior puede esconder que un
# modelo gane en la primera componente y pierda en las demas, que es lo que
# ocurre cuando el espectro FPCA es muy desigual.
filas_k = []
for nom, Yp in PRED.items():
    for etq, mask in (("train", es_train), ("test", ~es_train)):
        r = rmse_por_coeficiente(Y_obs[mask], Yp[mask])
        filas_k.append({"modelo": nom, "bloque": etq,
                        **{f"FPC{COMPONENT_IDX[k] + 1}": float(r[k])
                           for k in range(n_components)}})
rmse_k_df = pd.DataFrame(filas_k).set_index(["bloque", "modelo"]).sort_index()
rmse_k_df.to_csv(PATHS["out_report"] / f"{PREFIJO + 1}_rmse_por_componente.csv")

display(rmse_k_df.loc["test"].style.format("{:.4f}")
        .background_gradient(cmap="RdYlGn_r", axis=0)
        .set_caption("RMSE por componente FPCA · bloque de PRUEBA"))
```

## 5. Ventana móvil del error funcional

Lo que pide `§03_06`: la evolución del error recorriendo entrenamiento y
prueba, no una sola cifra por bloque. Ningún modelo se reentrena; lo que se
desliza es la ventana de *evaluación*. Las ventanas que cruzan $T_0$ mezclan
dentro y fuera de muestra y se dibujan punteadas.

En el Escenario 1 —estacionario— las curvas deberían ser **planas a ambos
lados** y el orden entre modelos estable a lo largo del eje. Un cruce de líneas
dentro del test indica que la ventaja de un modelo depende del tramo, que es
información que la tabla de §4 promedia y pierde.

```python
W_COMP = VENTANAS_W[len(VENTANAS_W) // 2]   # el mismo ancho de referencia de 11_04
print(f"ancho de ventana: w={W_COMP}  (de {VENTANAS_W})")

tablas_mod = []
for nom, Yp in PRED.items():
    t_ = ventana_movil_funcional(X_true_ev, curvas_desde_scores_std(Yp), grilla,
                                 T0_orig, w=W_COMP, t_offset=N_LAGS, verbose=False)
    t_["modelo"] = nom
    tablas_mod.append(t_)

t_trunc = ventana_movil_funcional(X_true_ev, X_proj, grilla, T0_orig,
                                  w=W_COMP, t_offset=N_LAGS, verbose=False)
t_trunc["modelo"] = "(truncamiento FPCA)"
tablas_mod.append(t_trunc)

# El PSBPM-FD se anade desde la tabla que escribio 11_04, si existe con este w.
# No se recalcula: seria la misma cifra tras releer trazas y propagarlas.
f_vent_psbp = PATHS["out_report"] / f"57_ventana_funcional_w{W_COMP}.csv"
if f_vent_psbp.exists():
    t_psbp = pd.read_csv(f_vent_psbp)
    t_psbp["modelo"] = "PSBPM-FD"
    tablas_mod.append(t_psbp[["t_centro", "bloque", "cruza_T0", "mise", "modelo"]])
    print(f"OK ventana movil del PSBPM-FD anadida desde {f_vent_psbp.name}")
else:
    print(f"AVISO falta {f_vent_psbp.name}: la figura queda sin el PSBPM-FD.")

ventana_df = pd.concat(tablas_mod, ignore_index=True)
ventana_df.attrs["w"] = W_COMP
ventana_df.to_csv(
    PATHS["out_report"] / f"{PREFIJO + 2}_ventana_modelos_w{W_COMP}.csv", index=False)
print(f"tabla de ventana movil: {ventana_df.shape[0]} filas "
      f"({len(tablas_mod)} series x {len(t_trunc)} ventanas)")
```

## 6. Contraste con el PSBPM-FD

Las cifras del modelo propuesto se **leen** de los CSV que escribió `11_04`; no
se recalculan. Si esos archivos no están, la celda lo dice y el resto del
notebook sigue siendo válido como comparación entre referencias.

La cifra que decide el eje 1 es el **MISE en el bloque de prueba**. El RMSE en
scores se reporta al lado porque un modelo puede ganar en scores y perder en la
curva cuando el error se concentra en las componentes de mayor autovalor.

```python
f_psbp_s = PATHS["out_report"] / "50_metricas_scores.csv"
f_psbp_f = PATHS["out_report"] / "51_metricas_funcionales.csv"

psbp = {}
if f_psbp_s.exists() and f_psbp_f.exists():
    _s = pd.read_csv(f_psbp_s, index_col=[0, 1])
    _f = pd.read_csv(f_psbp_f, index_col=0)
    for b in ("train", "test"):
        psbp[b] = {"RMSE_scores": float(_s.loc[b, "RMSE"].mean()),
                   "R2_scores":   float(_s.loc[b, "R2"].mean()),
                   "MISE":        float(_f.loc[b, "MISE"]),
                   "RMSE_func":   float(_f.loc[b, "RMSE_f"])}
    print("OK cifras del PSBPM-FD leidas de 50_/51_ (11_04)")
    print(f"   MISE test = {psbp['test']['MISE']:.6f}")
else:
    print("FALTAN 50_metricas_scores.csv y/o 51_metricas_funcionales.csv.")
    print("Ejecuta 11_04_evaluacion; §6 queda sin la fila del PSBPM-FD.")
```

```python
if psbp:
    filas = []
    for b in ("train", "test"):
        filas.append({"bloque": b, "modelo": "PSBPM-FD", **psbp[b]})
        for nom in PRED:
            r = comp_df.loc[(b, nom)]
            filas.append({"bloque": b, "modelo": nom,
                          "RMSE_scores": float(r["RMSE_scores"]),
                          "R2_scores": float(r["R2_scores"]),
                          "MISE": float(r["MISE"]),
                          "RMSE_func": float(r["RMSE_func"])})
        filas.append({"bloque": b, "modelo": "(truncamiento FPCA)",
                      "RMSE_scores": np.nan, "R2_scores": np.nan,
                      "MISE": MISE_TRUNC[b], "RMSE_func": np.nan})

    final_df = pd.DataFrame(filas).set_index(["bloque", "modelo"])
    # La primera fila de cada bloque es, por construccion, la del PSBPM-FD.
    ref_bloque = final_df.groupby("bloque")["MISE"].transform(lambda s: s.iloc[0])
    final_df["MISE_rel_PSBP"] = final_df["MISE"] / ref_bloque
    final_df.to_csv(PATHS["out_report"] / f"{PREFIJO + 4}_tabla_final.csv")

    display(final_df.style
            .format({"RMSE_scores": "{:.4f}", "R2_scores": "{:.4f}",
                     "MISE": "{:.6f}", "RMSE_func": "{:.6f}",
                     "MISE_rel_PSBP": "{:.3f}x"}, na_rep="—")
            .background_gradient(subset=["MISE"], cmap="RdYlGn_r")
            .set_caption("Tabla del eje 1 · MISE contra la curva VERDADERA · "
                         "MISE_rel_PSBP < 1 significa mejor que el PSBPM-FD"))
```

```python
if psbp:
    print("Veredicto sobre el bloque de PRUEBA (MISE, menor es mejor)\n")
    ref = psbp["test"]["MISE"]
    orden = (final_df.loc["test", "MISE"]
             .drop(index=["(truncamiento FPCA)"]).sort_values())
    for i, (nom, v) in enumerate(orden.items(), 1):
        marca = "   <- PSBPM-FD" if nom == "PSBPM-FD" else ""
        print(f"  {i}. {nom:22s} MISE={v:.6f}   ({v / ref:.3f}x){marca}")

    mejores = [n for n, v in orden.items() if v < ref]
    print()
    if not mejores:
        print("El PSBPM-FD domina en MISE a todas las referencias en el test.")
    else:
        print(f"Superan al PSBPM-FD en MISE: {', '.join(mejores)}")
        print("En el Escenario 1 —FAR(1) lineal gaussiano— que el VAR o el FAR")
        print("empaten o ganen NO es un fracaso: es el escenario correctamente")
        print("especificado para ellos, y el precio de la flexibilidad. La")
        print("pregunta del estudio es que ocurre en los escenarios 2-4.")

    _piso = MISE_TRUNC["test"] / ref
    print(f"\nEl piso de truncamiento es {_piso:.1%} del MISE del PSBPM-FD en test."
          + ("\n  <- la representacion domina el error; subir M rinde mas que "
             "cambiar de modelo" if _piso > 0.7 else ""))
```
